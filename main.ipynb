```python
# ── Resume helper ────────────────────────────────────────────────────────────
# Re-run this after reconnecting to find the latest checkpoint and print a
# ready-to-paste resume command.
import glob

train_ckpts = sorted(glob.glob(os.path.join(GDRIVE_EXPS, 'findkg-j*', 'Translator*.ckpt')))
rules_files = sorted(glob.glob(os.path.join(GDRIVE_EXPS, 'findkg-rule-*', 'rules.txt')))

latest_ckpt  = train_ckpts[-1] if train_ckpts  else None
latest_rules = rules_files[-1] if rules_files  else None

print('Latest checkpoint :', latest_ckpt  or 'none found')
print('Latest rules file :', latest_rules or 'none found')

skip_flags = []
if latest_ckpt:   skip_flags.append(f'--skip_train --ckpt {latest_ckpt}')
if latest_rules:  skip_flags.append(f'--skip_decode --rules_file {latest_rules}')

print(f'''
── Suggested resume command ──────────────────────────────────────
!python3 run_symbolic.py \\\
    --data_dir FinDKG_dataset --dataset FinDKG \\\
    --output {GDRIVE_OUTPUT} \\\
    --jump 2 --padding 200 --maxN 100 --epochs 50 \\\
    --exps_dir {GDRIVE_EXPS} \\\
    {' '.join(skip_flags)}
''')
```
